## UMAPnearestNeighbors
required - one UMAP Koordinate file and and reference annotation

NanoDiP_all_in_one_50 running as a server on the local machine will create *pie.csv for each sample

```python
def UMAPnearestNeighbors(Working_MSAsamplSheetDFi):
    
    i=0
    
    for row in Working_MSAsamplSheetDFi.itertuples(index=True, name="row"):
    
        ConcatenatedName = row.Sample_Name_InputM_Beta
        
        refAnno="ND_IfP_20250912.xlsx"
        
        #create UMAP plot for each sample 
        
        "http://localhost:8081/umapplot?sampleName=B2020_33674_20250929_R10&refAnno=GSE90496_IfP02.xlsx"
        
        curlCMD = 'curl "http://localhost:8081/umapplot?sampleName=' + ConcatenatedName + '&refAnno=' + refAnno+'"'
        print(curlCMD)
        
        p = subprocess.Popen(curlCMD, shell=True, stdout=subprocess.PIPE)
        p.wait() # wait until process is completed
        print(p)        
      
        pdfURL='curl "http://localhost:8081/makePdf?sampleName=' + ConcatenatedName + '&refAnno=' + refAnno +'"'
        print(pdfURL)
        p2=subprocess.Popen(pdfURL, shell=True, stdout=subprocess.PIPE)
        p2.wait()
        print(p2)           
```
